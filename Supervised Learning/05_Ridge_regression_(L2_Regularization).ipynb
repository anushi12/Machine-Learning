### Ridge regression

Ridge says: "Don't allow coefficients to become too large."

* So we add a penalty when coefficient grow. This is called regularization.

In ridge, the loss is calculated as -
  - Cost function = 1/2m ∑i=1n​(yi​−y^​i​)2+λ∑j=1p​βj2​
  - loss = squared error + penalty term

Lambda controls the penalty strength.
  - if lambda is high, then model becomes almost flat.
  - if lambda is small, then coeffients are slightly small.
  - if lambda is zero, then it's a normal linear regression.

Ridge formula -
    β=(XTX+λI)^−1XTy


```python
import numpy as np
import matplotlib.pyplot as plt

# create a dataset
X = np.array([
    [1,2],
    [2,4],
    [3,6],
    [4,8],
    [5,10]
])

y = np.array([5, 8, 14, 16, 23])

# add bias
X_b = np.c_[np.ones((X.shape[0],1)),X]
lambda_ = 1

n_features = X_b.shape[1]

# identity matrix
I = np.eye(n_features)
I[0,0] = 0 # we do not penalize the intercept

# Ridge formula
beta = np.linalg.inv(X_b.T.dot(X_b) + lambda_ * I).dot(X_b.T).dot(y)

print("coeffients: ", beta)
y_pred = X_b.dot(beta)
print("predictions: ", y_pred)

# Visualization
plt.scatter(X[:,0], y, label="Actual Data")
plt.plot(X[:,0], y_pred, color="red", label="Ridge Prediction")

plt.xlabel("Feature 1")
plt.ylabel("Target")
plt.legend()
plt.show()


```
